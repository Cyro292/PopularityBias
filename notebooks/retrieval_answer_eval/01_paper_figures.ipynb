# Retrieval Answer Evaluation

A concise, script-backed analysis of whether gold answer aliases occur in retrieved chunks. Retrieved Wikipedia IDs are not used to determine correctness.

```python
from pathlib import Path
import subprocess
import sys

REPO_ROOT = Path.cwd().resolve().parents[1] if Path.cwd().name == 'retrieval_answer_eval' else Path.cwd().resolve()
RESULTS_DIR = REPO_ROOT / 'data' / 'wiki_full_bil' / 'all_qa_8k'
PAPER_FIGURES = REPO_ROOT / 'paper_figures'
```

```python
subprocess.run([
    sys.executable, '-m', 'src.process.pipeline.retrieval_answer_eval_runner',
    '--output-dir', RESULTS_DIR.name,
    '--only-keys', 'bm25_plus', 'ivfpq_high',
], cwd=REPO_ROOT, check=True)

subprocess.run([
    sys.executable, '-m', 'src.process.analysis.plot_answer_retrieval_by_decile',
    '--results-dir', str(RESULTS_DIR),
    '--output-dir', str(PAPER_FIGURES),
], cwd=REPO_ROOT, check=True)
```
