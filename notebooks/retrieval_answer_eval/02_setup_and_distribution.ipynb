```python
reference = results_by_strategy[ALL_STRATEGIES[0]]
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
decile_counts = reference.groupby('decile').size().reindex(range(10), fill_value=0)
axes[0].bar(decile_counts.index + 1, decile_counts.values, color='#4C78A8')
axes[0].set(title='Evaluable Questions by Popularity Decile', xlabel='Popularity Decile (1=Rare, 10=Famous)', ylabel='Questions')
dataset_counts = reference['dataset'].value_counts().sort_values()
axes[1].barh(dataset_counts.index.str.replace('_', ' ').str.title(), dataset_counts.values, color='#59A14F')
axes[1].set(title='Evaluable Questions by Dataset', xlabel='Questions')
for ax in axes: ax.spines[['top', 'right']].set_visible(False)
fig.suptitle('Answer-Containment Evaluation Cohort', fontweight='bold')
fig.tight_layout()
output_path = IMAGES_DIR / 'answer_distribution_overview.png'
fig.savefig(output_path, dpi=300, bbox_inches='tight')
plt.show()
output_path
```
